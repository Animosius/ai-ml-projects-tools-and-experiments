```python
# Converting into 100 rows and column
transformed_data = original_data.drop(columns=["Year", "Soil type", "Crops", "Location", "Season"]).iloc[:100]
transformed_data.rename(columns={"yeilds":"Yield"}, inplace=True)
transformed_data.head()
```

## Relationship

```python
# Area and yield
plt.scatter(data["Area"], data["Yield"])
plt.title("Area vs Yield")
plt.xlabel("Area (ha)")
plt.ylabel("Yield (kg)")
plt.show()

corr_value = float(data["Area"].corr(data["Yield"]))
print(f"Correlation value {corr_value}")
```

```python
# Rainfall and yield
plt.scatter(data["Rainfall"], data["Yield"])
plt.title("Rainfall vs Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (kg)")
plt.show()

corr_value = float(data["Rainfall"].corr(data["Yield"]))
print(f"Correlation value {corr_value}")
```

```python
# Temperature and yield
plt.scatter(data["Temperature"], data["Yield"])
plt.title("Temperature vs Yield")
plt.xlabel("Temperature (celcius)")
plt.ylabel("Yield (kg)")
plt.show()

corr_value = float(data["Temperature"].corr(data["Yield"]))
print(f"Correlation value {corr_value}")
```

```python
# Humidity and yield
plt.scatter(data["Humidity"], data["Yield"])
plt.title("Humidity vs Yield")
plt.xlabel("Humidity (percentage)")
plt.ylabel("Yield (kg)")
plt.show()

corr_value = float(data["Humidity"].corr(data["Yield"]))
print(f"Correlation value {corr_value}")
```

```python
# Yield and Price
plt.scatter(data["Yield"], data["price"])
plt.title("Yield vs Price")
plt.xlabel("Yield (kg)")
plt.ylabel("Price")
plt.show()

corr_value = float(data["Yield"].corr(data["price"]))
print(f"Correlation value {corr_value}")
```
