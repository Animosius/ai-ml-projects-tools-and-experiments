# 1. Introduction to scikit-learn

What is scikit-learn ?

In a nutshell scikit-learn is a python library dedicated to AI and Machine Learning !

So why the 'learn' in 'scikit-learn' ? doesn't that mean this is for newbies ?

Not at all actually ! The 'learn' comes because this library is essentially a toolbox that helps you make systems that learns.

In machine learning (in scikit-learn), **learning** comes in a few step:

1. Getting data
2. Dataset handling and preprocessing
3. Selecting / building our model that fits the data
4. Fitting the model 
5. Do Inference with the model
