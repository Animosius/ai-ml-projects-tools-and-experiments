# 3. Dataset Handling and Preprocessing 

What is a dataset ?

A dataset is essentially an amalgamation of data, usually for the sake of training and testing a model



For the more advanced:

What datatype do most models in scikit-learn take ?

Mostly 2D datatypes.

eg.
[[1],[2],[3], [4],[5]]

Nonetheless, even if the data is in the form of a numpy or pandas object, it is still acceptable.

packages used -> `numpy` & `pandas` & `sklearn.model_selection.train_test_split`

In supervised learning, a dataset can be split into features and labels.

Features -> Variables used to make predictions (usually denoted as X)

Labels -> Variables we want to predict (usually denoted as y)

```python
# Let's make a dataset, let's use an agricultural themed dataset for simplicity  

import pandas as pd

agri_data = {
    "rainfall": [1, 2, 4, 5, 8, 11, 13, 14 , 17, 19],
    "fertilizer": [3, 5, 6, 7, 8, 10, 12, 15, 16, 19],
    "moss_growth": [20, 25, 27, 31, 33, 34, 37, 39, 43, 44]
}

df = pd.DataFrame(agri_data) # Now we turn the data into a pandas dataframe file format
```

```python
# Check data
df
```

```python
# Now let's split the data into features and labels

X = df[["rainfall", "fertilizer"]] # This is our features 
y = df["moss_growth"] # This is our labels
```

Train test_test_split:

In machine learning datas are separated into train and testing batch:

Train batch:
1) Features for training (usually denoted as X_train)
2) Labels for training (usually denoted as y_train)

Test batch:
1) Features for testing (usually denoted as X_test)
2) Labels for testing (usually denoted as y_test)

Usually, the ratio between test and train dataset are:

* 6:4
* 7.5:2.5
* 8:2

Fortunately, sklearn has provided us with a way to do this easily by using:

`sklearn.model_selection.train_test_split`

```python
# Splitting our dataset

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=67
)
```

```python
# Now let's see what our features look like
X_train, X_test
```

```python
# Then our labels
y_train, y_test
```

```python
# Now that we have the dataset, let's fit a model shall we ?

from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train, y_train)
```

```python
# Cool ! Now let's do some testing

linear_model.predict(X_test)
```

```python
# But...what if we want to know what our accuracy score is ?
# for a linear model we use an r2 score

from sklearn.metrics import r2_score

y_pred = linear_model.predict(X_test)

score = r2_score(y_test,y_pred) # r2-score(actual, predicted)
print(f"R2 score: {score}")
```

```python
# or a cleaner way to do it 

linear_model.score(X_test, y_test)
```

# 4. Regression 

What is regression ?

Regression = **learning a function** that maps inputs -> outputs from data

Given: (X, y)
Find: f such that y = f(x)

Regression in machine learning is where we get a set of values of data and try to predict what is the next value by approximating a function.

from X & y dataset -> approximate f(X) -> input new X -> get y_pred 

Different types of methods for regression
1) Linear methods -> linear regression (Ordinary least square method (minimises (y-y_pred)^2))
2) Ensemble methods -> RandomForestRegressot, Gradient boost

packages used -> `sklearn.linear_models` & `sklearn ensemble`

## 4.0 Dataset making and handling

```python
# Let's make a dataset 

import pandas as pd
import numpy as np

np.random.seed(42)

n = 100

data = {
    "mol_weight": np.random.normal(300, 50, n),
    "logP": np.random.normal(2.5, 1.0, n),
    "h_donors": np.random.randint(0, 6, n),
    "h_acceptors": np.random.randint(2, 12, n),
    "tpsa": np.random.normal(75, 20, n),
    "rotatable_bonds": np.random.randint(0, 10, n),
}

df = pd.DataFrame(data)

# Create a semi-realistic target
df["bioactivity_score"] = (
    0.02 * df["mol_weight"]
    + 1.5 * df["logP"]
    - 1.2 * df["h_donors"]
    - 0.8 * df["h_acceptors"]
    - 0.01 * df["tpsa"]
    - 0.5 * df["rotatable_bonds"]
    + np.random.normal(0, 2, n)  # noise
)

df.head()
```

```python
# Now we split the data into features and labels
from sklearn.model_selection import train_test_split

X = df.drop("bioactivity_score", axis=1) 
y = df["bioactivity_score"]

# for X we're dropping the target variable from the features dataset
# we put axis=1 because we want to drop a column, if it was axis=0 it would drop a row instead

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=69
)
```

```python
# Checking the data split shapes

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

The goal of the data is to predict the bioactivity score of a molecule based on given features

So this is a regression problem because we need to generate a value as prediction

## 4.1 Linear methods for regression

```python
# Now let's see which model fits this 

from sklearn.linear_model import LinearRegression

bio_lrm = LinearRegression() # bio_lrm -> bioactivity linear regression model
bio_lrm.fit(X_train, y_train)
```

```python
# Now we check the parameter of the model

print(f"Model coefficient: {bio_lrm.coef_}")
print(f"Model bias: {bio_lrm.intercept_}")
```

```python
# Let's interpret it:

print(f"Molecular weight coeff; {bio_lrm.coef_[0]} {"Strongest_factor" if bio_lrm.coef_[0] == max(bio_lrm.coef_) else "Not_strongest_factor"}")
print((f"Molecular logP coeff: {bio_lrm.coef_[1]} {"Strongest_factor" if bio_lrm.coef_[1] == max(bio_lrm.coef_) else "Not_strongest_factor"}"))
print(f"Molecular H_donor coeff: {bio_lrm.coef_[2]} {"Strongest_factor" if bio_lrm.coef_[2] == max(bio_lrm.coef_) else "Not_strongest_factor"}")
print(f"Molecular H_acceptor coeff: {bio_lrm.coef_[3]} {"Strongest_factor" if bio_lrm.coef_[3] == max(bio_lrm.coef_) else "Not_strongest_factor"}")
print(f"Molecular TPSA coeff: {bio_lrm.coef_[4]} {"Strongest_factor" if bio_lrm.coef_[4] == max(bio_lrm.coef_) else "Not_strongest_factor"}")
print(f"Molecular Rotatable Bonds coeff: {bio_lrm.coef_[5]} {"Strongest_factor" if bio_lrm.coef_[5] == max(bio_lrm.coef_) else "Not_strongest_factor"}")
```

```python
# A better way

import numpy as np

coefs = bio_lrm.coef_
max_idx = np.argmax(np.abs(coefs))

for i, coef in enumerate(coefs):
    label = "strongest_factor" if i == max_idx else "not_strongest_factor"
    print(f"Feature {i}: {coef:.3f} -> {label}")
```

```python
# To sort things out 
sorted_idx = np.argsort(np.abs(coefs))[::-1]

for i in sorted_idx:
    print(f"Feature {i}: {coefs[i]:.3f}")
```

```python
# Now let's check the R^2 score of our model

print(f"R2-score: {bio_lrm.score(X_test, y_test)}")
```

What does this mean ?

R2-score: ~77%

This means that that our model only explains ~77% variance of our data.

Remainning ~23% = noise / missing features / non-linearity
